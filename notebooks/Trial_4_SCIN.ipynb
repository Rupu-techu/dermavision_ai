```python
from datasets import load_dataset

scin = load_dataset(
    "google/scin",
    split="train",
    streaming=True
)

print(scin)
```

```python
# Get the first record from the streaming dataset

first_record = next(iter(scin))

# Print all columns and their values
for column, value in first_record.items():
    print(f"{column}: {value}")
```

```python
print("Total number of features:", len(first_record))
```

```python
label_columns = [
    column for column in first_record.keys()
    if "dermatologist_skin_condition_label" in column
]

print(label_columns)
```

```python
important_columns = [
    "age_group",
    "sex_at_birth",
    "fitzpatrick_skin_type",
    "condition_duration"
]

for column in important_columns:
    print(column, ":", first_record.get(column))
```

```python
label_columns = [
    column for column in first_record.keys()
    if "dermatologist" in column.lower()
]

print("Dermatologist-related columns:")
for column in label_columns:
    print(column)
```

```python
skin_columns = [
    column for column in first_record.keys()
    if "skin_condition" in column.lower()
]

print("Skin condition-related columns:")
for column in skin_columns:
    print(column)
```
